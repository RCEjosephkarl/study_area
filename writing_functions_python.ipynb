## Chapter 01: Best Practices

### Docstrings
Docstring - the first description of the function. Uses triple quotes(""" ___ """) as this docstring uses mulitple lines.

```python
# Sample function with docstring in complete NumpyDoc format

def function_name(arguments):
    """
    Description of what the function does
    
    Description of each argument //Args//
    
    Description of the return value(s) //return//
    
    Description of errors raised, if any //raises//
    
    Optional extra notes or examples of usage. //notes//
    """
```

Use `.__doc__` attribute for each function to access the docstring when executing.

```python
# Using .__doc__ attribute

def print_five():
    """Output the value of 5
    
    Returns:
    5 (int)
    """
    five = 5
    return five

print(print_five())
print(print_five.__doc__)

#
```

### Don't repeat yourself (DRY)
![don'tRepeatYourself](DRY.png)

- It is too confusing with repetitive functions with different arguments.
- Vague codes like these will make the bug-finding more difficult.
- Instead of these repetitions, create a function to cover these repetitive commands. However, this is still not enough.



### Do One Thing
![doOneThing](doOneThing.png)
- Every single function should have only one responsibility; one formula.

**Advantages of doing one thing**
- More flexibility
- Easier to comprehend
- Simpler to test
- Simpler to debug
- Easier to modify/alter.

## Chapter 02: Context Managers

### Using a context manager

There are two ways two define a context manager:
1. Class-based 
2. **_ Function-based_**



```python
# Using a context manager

with open('df.txt' , 'r') as data:
    tab = data.read()
    length = len(tab)
    
print('The file is {} characters long.'.format(length))
print('-----------------------------------------------------------')
print(tab)
```

### **Function-based context manager**

Steps:
1. Define a function
2. Add any set up code your context needs. (optional)
3. Use the `yield` keyword.
4. Add any teardown code your context needs.
5. Add the `@contextlib.contextmanager` decorator.
