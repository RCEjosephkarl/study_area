### _**Transforming data in a pipeline**_

If data is curated in an incorrect format, or information is lost through the transformation process, downstream users may not be able to generate value from the output of a data pipeline. `pandas` provides powerful tools to transform tabular data. `.loc[]` and .`to_datatime()` methods will be used to transform data.


- `.loc[]` allows for both dimensions of a DataFrame to be transformed.

        # Keep only non-zero entries
        cleaned = raw_stock_data.loc[raw_stock_data["open"] > 0, :]
        
        # Remove excess columns
        cleaned = raw_stock_data.loc[:, ["timestamps", "open", "close"]]
        
        # Combining into one step
        cleaned = raw_stock_data.loc[raw_stock_data["open"] > 0, ["timestamps", "open", "close"]]

- `.iloc[]` uses integer indexing rather than loc's label-based and boolean indexing.

        cleaned = raw_stock_data.iloc[[0:50],[0,1,2]]

### _Persisting data in an ETL Pipeline_

Loading data to a file:
- Ensures data consumers have stable access to transformed data
- Occurs as a final step in an ETL process, as well as between discrete steps
- Captures a "snapshot" of the data

`to_csv()` method

        import pandas as pd
        
        # Data extraction and transformation
        raw_data = pd.read_csv("raw_stock_data.csv")
        stock_data = raw_data.loc[raw_data["open"] > 100, ["timestamps", "open"]]
        
        # Load data to a .csv file
        stock_data.to_csv("stock_data.csv")
        
- `.to_csv()` is called on the DataFrame
- Writes DataFrame to path "stock_data.csv


**Customizing CSV file output**

        stock_data.to_csv("./stock_data.csv", header = _____, index = _____ , sep = "|")
        
- for `header` parameter, expected of _boolean_ value or a list `[]` of string values
- for `index` parameter, expected of _boolean_ value if index column is written to the file
- for `sep` parameter, default is the comma string `,`, the pipe `|` could be a common option


_relative counterparts:_

- `.to_parquet()`
- `.to_json()`
- `.to_sql()`
