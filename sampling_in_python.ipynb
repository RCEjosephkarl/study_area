## Stratified and Weighted Random Sampling

### Stratified technique

- a technique that allows us to sample a population that contains subgroups

```python
# Find the most recorded device models
top_device = data['Device Model'].value_counts()
print(top_device.head(3))

top_device_list = list(top_device.head(3).index)

print(top_device_list)
```

```python
#Subsetting the population
top_device_subset = data['Device Model'].isin(top_device_list)
print(data[top_device_subset])
```

```python
#Simple Random Sampling of the entire dataset
print(data['Device Model'].value_counts(normalize = True))

#Simple Random Sampling of 10% of the subset
top_device_samp = data[top_device_subset].sample(frac = 0.1, random_state = 2024)
print(top_device_samp['Device Model'].value_counts(normalize = True))
```

**My Comment** : The differences in proportionality does not make sense.

```python
#Equal counts stratified sampling

data_eq = data.groupby('Device Model').sample(n=15, random_state = 101)
print(data_eq)
```

```python
#Equal counts stratified sampling (Part 2)
# Test the proportionality of the sample
print(data_eq['Device Model'].value_counts(normalize = True))

```

### Weighted Random Sampling

- Specify weights to adjust the relative probability of a row being sampled.

```python
#Example of weighted random sampling

import numpy as np
weight_device_samp = top_device_samp
condition = weight_device_samp['Device Model'] == 'Xiaomi Mi 11'
# Tag 2 if  it matches the condition, else tag 1
weight_device_samp['weight'] = np.where(condition,2,1)

weight_device_samp = weight_device_samp.sample(frac = 0.1, weights = 'weight')
print(weight_device_samp)
```

## Cluster Sampling

One problem with stratified sampling is we need to collect data from every
subgroup. Case(s) similar to these are usually time-expensive and resource-expensive.

**Stratified Sampling versus Cluster Sampling**

_Stratified Sampling_
 - Split the population into subgroups
 - Use simple random sampling on every subgroup

_Cluster Sampling_
 - Use simple random sampling to pick some subgroups
 - Use simple random sampling on only those subgroups



```python
#Varieties of coffee
phone_models_pop = list(data['Device Model'].unique())
print(phone_models_pop)
```

```python
#Cluster Sampling (Stage 1)
import random
phone_models_clsamp = random.sample(phone_models_pop, k = 2)

#Cluster Sampling (Stage 2)
data['Device Model'] = data['Device Model'].astype('category')
model_condition = data['Device Model'].isin(phone_models_clsamp)
cluster_model = data[model_condition]
cluster_model['Device Model'] = cluster_model['Device Model'].cat.remove_unused_categories()

cluster_model.groupby('Device Model')\
    .sample(n = 5 , random_state = 2024)
cluster_model.head()
```

## Relative Error of Point Estimates
- Sample size = number of rows

    **_Relative Error_** is the absolute value in difference between population parameter and a point estimate.
    - Population Parameter for example, population mean
    - Point Estimate for example, sample mean of _n_ samples

```python
#Relative Error in percentage
population_mean = data['Battery Drain (mAh/day)'].mean()
sample_mean = data['Battery Drain (mAh/day)'].sample(n =90).mean()
rel_error_pct = 100 * abs ((population_mean - sample_mean)/population_mean)
print(len(data))
print(rel_error_pct)
```

## Exact and Approximate Sampling Distribution


```python
import itertools
import pandas as pd

# Define the expand_grid function
def expand_grid(dictionary):
    rows = itertools.product(*dictionary.values())
    return pd.DataFrame.from_records(rows, columns=dictionary.keys())

# For ______ Sampling Distribution
# Expand a grid representing 5 8-sided dice
dice = expand_grid({
    'die1' : [1,2,3,4,5,6,7,8],
    'die2' : [1,2,3,4,5,6,7,8],
    'die3' : [1,2,3,4,5,6,7,8],
    'die4' : [1,2,3,4,5,6,7,8],
    'die5' : [1,2,3,4,5,6,7,8],
})

# Print the result
print(dice)
```

```python
# For Approximate Sampling Distribution
# Simulating means of four dice rolls

import numpy as np
sample_means_1000 = []

for i in range(1000):
    sample_means_1000.append(
        np.random.choice(list(range(1,7)), size = 4, replace=True).mean()
    )

print(sample_means_1000[100:111])
```

## The Central Limit Theorem and the Standard Errors

- As the sample size increases, the sample mean value approaches to the population mean value.
- As the sample size increases, the distribution of the averages gets closer to being normally distributed.
- As the sample size increases, the width of the sampling distribution gets narrower.
- As the sample size increases, the _Standard Deviation_ of the sample means decreases.
